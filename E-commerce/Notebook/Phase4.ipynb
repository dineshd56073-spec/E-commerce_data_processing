```python
import pandas as pd
import numpy as np
df=pd.read_csv("online_retail_practice_sample.csv")
df
```

```python
df_raw=df.copy()
df_raw=df_raw.rename(columns={
    'Description':'ProductDescription'
})
df_raw
```

```python
df_raw['Revenue'] = df_raw['Quantity'] * df_raw['UnitPrice']
df_raw
# print(df_raw[['Quantity', 'UnitPrice', 'Revenue']])
```

```python
con = [
    df_raw['InvoiceNo'].str.startswith('C', na=False),
    df_raw['Quantity'] < 0,
    (df_raw['Quantity'] > 0) & (df_raw['UnitPrice'] >= 0)
]

choices = ['Cancellation', 'Return', 'Positive Sale']

df_raw['TransactionType'] = np.select(
    con,
    choices,
    default='Other/Invalid'
)
df_raw
```

```python
df_raw['IsIdentifiedCustomer'] = df_raw['CustomerID'].notna()
df_raw
```

```python
df_raw['InvoiceDate']=pd.to_datetime(
    df_raw['InvoiceDate']
)
df_raw
```

```python
df_raw['InvoiceNo'] = df_raw['InvoiceNo'].astype('string')
df_raw['StockCode'] = df_raw['StockCode'].astype('string')
df_raw['Country'] = df_raw['Country'].astype('string')
df_raw.dtypes
```

```python
df_raw=df_raw.sort_values(by=['InvoiceDate','InvoiceNo'],ascending=[True,True])
df_raw
```
