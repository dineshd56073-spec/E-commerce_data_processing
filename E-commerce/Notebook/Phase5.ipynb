```python
import pandas as pd
import numpy as np
df=pd.read_csv("online_retail_practice_sample.csv")
df
```

```python
df_raw=df.copy()
missing_count = df_raw.isnull().sum()
missing_percentage = (missing_count / len(df_raw)) * 100

missing_summary = pd.DataFrame({
    'MissingCount': missing_count,
    'MissingPercentage': missing_percentage
})

print(missing_summary)
```

```python
missing_percentage = (df.isnull().mean() * 100)

ranked_columns = missing_percentage.sort_values(
    ascending=False
)

print(ranked_columns)
```

```python
missing_customer = df[df["CustomerID"].isnull()]

print(missing_customer)

print("Number of affected transactions:",
      len(missing_customer))
```

```python
description = df["Description"].astype("string")

condition = (
    description.isna()
    | description.str.strip().eq("").fillna(False)
)

missing_description = df[condition]

print(missing_description)

print("Number of affected records:",
      len(missing_description))
```

```python
missing_report = pd.DataFrame({
    "ColumnName": df_raw.columns,
    "MissingCount": df_raw.isnull().sum().values,
    "NonMissingCount": df_raw.notnull().sum().values,
    "MissingPercentage": (
        df_raw.isnull().mean().values * 100
    )
})

print(missing_report)

missing_report.to_csv(
    "missing_value_report.csv",
    index=False
)
```

```python
duplicate_rows = df_raw[df_raw.duplicated(keep=False)]

duplicate_count = df_raw.duplicated().sum()

print("All rows belonging to duplicate groups:")
print(duplicate_rows)

print("Duplicate rows beyond first occurrences:",
      duplicate_count)

print("Total rows involved in duplicate groups:",
      len(duplicate_rows))
```

```python
df_clean = df.copy()

rows_before = len(df_clean)

df_clean = df_clean.drop_duplicates()

rows_after = len(df_clean)

print("Rows before removal:", rows_before)
print("Rows after removal:", rows_after)
print("Rows removed:", rows_before - rows_after)

print("Remaining duplicate rows:",
      df_clean.duplicated().sum())
```

```python
repeated_records = df_raw.merge(
    df_raw[["InvoiceNo", "StockCode"]],
    on=["InvoiceNo", "StockCode"],
    how="inner"
)

print(repeated_records)
```

```python
df_clean = df_raw.copy()

df_clean["Description"] = (
    df_clean["Description"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

print("Missing descriptions after cleaning:",
      df_clean["Description"].isna().sum())

print("Missing CustomerID values:",
      df_clean["CustomerID"].isna().sum())
```
